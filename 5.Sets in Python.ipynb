## 5. Sets in Python

---
1. Introduction of Sets

A `set` in Python is a collection data type used to store multiple unique elements.

It is based on the mathematical concept of sets.

##### Key Characteristics
1. Unordered
2. Unindexed
3. Mutable(can add/remove items)
4. Does not allows duplicates value
5. Elements must be immutable(hashable)


```python
# How to create empty set 

s = {} # bydefault it is dictionary
print(s)
print(type(s))
```

```python
s = {1,20}  # sets value is key 
print(s)
print(type(s))
```

### 4. Data Types allowed in Sets

Elements must be Immutable(hashable)

Not Allowed:

Mutable types:
* List

### 10. Set Operations (Most Important Points of set)

Python sets support mathematical operations

1. Union
2. Intersection
3. difference
4. symmetric_difference

```python
A = {1,2,3}
B = {3,4,5}
print(A)
print(B)
```

1. Union: Unique Value

Symbol- pipe(|)

method- union()

```python
A = {1,2,3}
B = {3,4,5}
print(A)
print(B)
```

```python
## symbol
print(A|B)
```

```python
# method 

print(A.union(B))
```

```python
A = {1,2,3}
B = {3,4,5}
print(A)
print(B)
```

## Intersection: Common Elements

Symbol-  Ampersand(&)

Method- .intersection()

```python
# symbol 
print(A&B)
```

```python
# method 
print(A.intersection(B))
```

### 3. Difference : Subtract 

symbol- [-]

method: .difference()

```python
A = {1,2,3}
B = {3,4,5}
print(A)
print(B)

# A-B
# B-A
```

```python
# symbol 
print(A-B)
print(B-A)
```

```python
# method 

print(A.difference(B))
print(B.difference(A))
```

### Symmetric_difference

Elements in either set but not both

Symbol: ^

method: .symmetric_difference()

```python
A = {1,2,3}
B = {3,4,5}
print(A)
print(B)
```

```python
# symbol 
print(A^B)
```

```python
print(B^A)
```

```python
print(A.symmetric_difference(B))
```

```python
print(B.symmetric_difference(A))
```

#### 11. Set Comparison Operations

Subset()

Check if set is subset()

```python
A = {1,2}  # less
B = {1,2,3}  # more
print(A.issubset(B))  # True
```

```python
print(B.issubset(A))
```

```python
A = {1,2}  # less
B = {1,2,3}  # more

print(A.issuperset(B))
```

```python
print(B.issuperset(A))
```

### disjoint

Check it not common elements

```python
A = {1,2}
B = {3,4}

print(A.isdisjoint(B))  # True
```

```python
A = {1,2,3}
B = {3,4}

print(A.isdisjoint(B))  # True
```

### 18. When to Use Sets

Use Sets when:

* ✅ Remove Duplication
* ✅ Fast Searching
* ✅ Mathematical Operations: Union, intersection, difference, symmetric_difference, frozenset, superset, subset, disjoint
* ✅ Membership testing

Example:

* Unique User IDs
* Unique Email Address
* Tag System
* Data filtering
