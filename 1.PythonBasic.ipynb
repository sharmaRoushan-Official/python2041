### Data Typecastring


```python
num = 10  # string,float
print(num,type(num))
strNum = str(num)
print(strNum,type(strNum))

floatNum = float(num)
print(floatNum,type(floatNum))
```

```python
bool1 = True   # string, float, integer
print("Original Value:",bool1)
print("Data Type:",type(bool1))

boolStr = str(bool1)
print(boolStr,type(boolStr))

boolFloat = float(bool1)
print(boolFloat,type(boolFloat))

boolInt = int(bool1)
print(boolInt, type(boolInt))

bool2 = False # string, float, integer
print("Original Value:",bool2)
print("Data Type:",type(bool2))
```
