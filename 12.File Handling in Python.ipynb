### 12. Reading Files in Python


Q. What is File Handling?
`File Handling` allows python program to read data file and write data files permently.

📌Why we need files:

* Stored data permently
* Read logs, report, CSV(Comma Seperated File),.xlsx
* Process large data
* Avoid manual input

#### Types of Files
1. Binary Files: [.jpg,.pdf,.exe]
2. Text Files:[.txt,.csv,.xlsx]

------
## Steps to Read a file in Python

* ✮1.Open the file
* ✮2.Read the file
* ✮3.Close the file

### Opening a File
`file = open("filePath","Mode")`

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

# step-1
file  = open(fileName,'r')

# step-2
content  = file.read()

print(content)

# step-3
file.close()
````

### Modes in file 

|Mode|Meaning|
|----|----|
|`r`|Read(default)|
|`w`|Write|
|`a`|append|
|`r+`|Read & Write|

### Method 1: Read() ---> Read Entire File

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

# step-1
file  = open(fileName,'r')

# step-2
content  = file.read()  # Entire data read

print(content)

# step-3
file.close()
````

Notes:

* Reads whole file at once
* No suitable for very large files.

#### Method 2: readline() --> Read One line at a time.



````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

# step-1
file  = open(fileName,'r')

# step-2
content  = file.readline() # first line
content2 = file.readline() # second line
content3 = file.readline() # third line
print(content)
print(content2)
print(content3)
# step-3
file.close()
````

Note:

Reads only one line at a time.

### Method 3: readlines() --> Read all lines as list



````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

# step-1
file  = open(fileName,'r')

# step-2
content  = file.readlines()

print(content)

# step-3
file.close()
````

### Syntax for Opening a file

`fileObject = Open("fileName","mode")`


### Parameters
----
|`Parameter`|`Description`|
|-----|----|
|`fileName`|Name of the file|
|`mode`|Operation to perform|


### 3. File Modes

|Mode| Description|
|----|-----|
|`r`|Read file(Default)|
|`w`|Write File(Overwrite)|
|`a`|Append data|
|`x`|Create file if exists ti throw error|
|`rb`|Read binary file|
|`wb`|Write binary|

#### Read file using Loop

Best method for large files.


````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"
file = open(fileName,'r')


for line in file:
    print(line)

file.close()
````

### Using with statement [Best Recommendation]

With automatically closes the file.

Syntax:

```
with open('fileName'.'Mode') as varaible:
    statement
```

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"


with open(fileName,'r') as file:
    print(file.read())
````

### Check file Exists or not.

````python
import os
````

````python
# help(os)
````

````python
fileName = "C:\\Users\\User\\Desktop\\NursaryRhymes.txt"

if os.path.exists(fileName):
    print("File Exists")
    with open(fileName,'r') as file:
        print(file.read())
else:
    print("File Not Exists")
````
