## 10. Exceptional Handling in Python

Exeptional handling in Python is used to handle runtimeError so that the program does not crash suddenly and can continue execution smoothly.

In real-world software development, exception handling is mandatory because errors can occurs anytime due to:

* wrong user input
* missing file
* database failure
* network issue
* division by zero
* invalid operation

### 7. Multiple Except Blocks

You can handle Multiple Errors



````python
num1 = int(input("Enter a number: "))
num2 = int(input("Enter another number: "))
print(num1/num2)
````

````python
try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter another number: "))
    print(num1/num2)
except ZeroDivisionError:
    print("ZeroDivisionError")
except ValueError:
    print("Value Error Occurred")
````

## 8. Single Except Hanling Multiple Exception.

````python
try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter another number: "))
    print(num1/num2)
except (ZeroDivisionError,ValueError):
    print("ZeroDivisionError and ValueError")
print("done")
````

###  10. try-except-else Block

Else Block runs if no exeption Occurs

````python
try:
    x = int(input("Enter first value: "))
    y = int(input("Enter second value: "))
    val = x/y
except Exception as e:
    print("Error Occurred:",e)
else:
    print(f"After division of two numbers {x} and {y} Result is:{val}")
````

Note:

* if exception --> except runs
* If no exception --> else runs


### 11. try-except-else-finally Block
finally block runs always(wheather exception happens or not)

````python
try:
    x = int(input("Enter first value: "))
    y = int(input("Enter second value: "))
    val = x/y
except Exception as e:
    print("Error Occurred:",e)
else:
    print(f"After division of two numbers {x} and {y} Result is:{val}")
finally:
    print("Program Finished")  # file closing ---> File Handling
````

Note:

* if file exists: - file content+ "Program finished"
* if file not exists --> file content not found + "Program Finished"
