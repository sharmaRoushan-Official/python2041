## 9. Functions in Python

A `function` is a reusable block of code that perform a specific task.

### 1. Why we use functions?
* Avoid code repetation
* Make code readable
* Improve Modularity
* Easier debugging
* Reusability

Syntax:

```
def FunctionName():
    block of code
```

### There are generally Three types of functions

1. Built-in function
2. User-Defined function
3. Anonymous function(lambda function)

**Math & Numeric**

- `abs()` — Returns the absolute value of a number.
- `divmod()` — Returns the quotient and remainder of a division.
- `max()` — Returns the largest item in an iterable.
- `min()` — Returns the smallest item in an iterable.
- `pow()` — Returns x to the power of y.
- `round()` — Rounds a number to a specified decimal precision.
- `sum()` — Sums the items of an iterable.

**Type Conversion**

- `bool()` — Converts a value to a Boolean (`True` or `False`).
- `bytearray()` — Returns a mutable array of bytes.
- `bytes()` — Returns an immutable bytes object.
- `complex()` — Creates a complex number.
- `float()` — Converts a number or string to a floating-point number.
- `int()` — Converts a number or string to an integer.
- `str()` — Converts an object to a string.

**Collections & Iterables**

- `dict()` — Creates a dictionary.
- `frozenset()` — Returns an immutable set object.
- `list()` — Creates a list.
- `range()` — Generates a sequence of numbers.
- `set()` — Creates a new set.
- `slice()` — Returns a slice object for slicing sequences.
- `tuple()` — Creates a tuple.

**Data Manipulation & Iteration**

- `all()` — Returns `True` if all elements in an iterable are true.
- `any()` — Returns `True` if any element in an iterable is true.
- `enumerate()` — Returns an enumerate object (index and value pairs).
- `filter()` — Filters elements from an iterable based on a function.
- `iter()` — Returns an iterator object.
- `len()` — Returns the length of an object.
- `map()` — Applies a function to all items in an input iterable.
- `next()` — Retrieves the next item from an iterator.
- `reversed()` — Returns a reversed iterator.
- `sorted()` — Returns a sorted list of an iterable's elements.
- `zip()` — Combines multiple iterables into tuples.

**Input/Output & Files**

- `input()` — Prompts the user for input and reads it as a string.
- `open()` — Opens a file and returns a file object.
- `print()` — Prints objects to the text stream or console.

**String & Character Representation**

- `ascii()` — Returns a readable string version of an object, escaping non-ASCII characters.
- `bin()` — Converts an integer to a binary string.
- `chr()` — Converts an integer Unicode code point to its character.
- `hex()` — Converts an integer to a hexadecimal string.
- `oct()` — Converts an integer to an octal string.
- `ord()` — Converts a character to its integer Unicode code point.
- `repr()` — Returns a printable string representation of an object.

**Object & Class Inspection**

- `callable()` — Returns `True` if the object appears callable (like a function).
- `classmethod()` — Transforms a method into a class method.
- `delattr()` — Deletes an attribute from an object.
- `dir()` — Returns a list of valid attributes for an object.
- `getattr()` — Gets the value of a named attribute from an object.
- `hasattr()` — Checks if an object has a specified attribute.
- `hash()` — Returns the hash value of an object.
- `id()` — Returns the unique identity/memory address of an object.
- `isinstance()` — Checks if an object is an instance of a specific class.
- `issubclass()` — Checks if a class is a subclass of another class.
- `object()` — Returns a new featureless base object.
- `property()` — Gets, sets, or deletes an attribute via properties.
- `setattr()` — Sets the value of an attribute on an object.
- `staticmethod()` — Transforms a method into a static method.
- `super()` — Returns a proxy object representing a parent class.
- `type()` — Returns the type of an object or creates a new type.

**Code Execution & Scopes**

- `compile()` — Compiles source code into an executable code object.
- `eval()` — Evaluates a dynamically parsed Python expression.
- `exec()` — Dynamically executes Python code block statements.
- `globals()` — Returns a dictionary of the current global symbol table.
- `locals()` — Returns a dictionary of the current local symbol table.
- `vars()` — Returns the `__dict__` attribute of an object.
- `__import__()` — An advanced function invoked by the `import` statement.

**Asynchronous Programming**

- `aiter()` — Returns an asynchronous iterator.
- `anext()` — Returns the next value of an asynchronous iterator.

**Advanced & Miscellaneous**

- `breakpoint()` — Drops you into the debugging console.
- `help()` — Invokes the built-in system helper documentation.
- `memoryview()` — Returns a memory view object to access an object's internal buffer.

## 2. Keyword Arguments


````python
def student_details(name,age):
    return f"Student name is: {name}\nAge is: {age}" # function closed
    print("Hello world")

obj1 = student_details(age=45,name="Rahul")  # right position Note: Always follows the position of parameter
print(obj1)
````

### 3. Default Arguments

---

````python
def student_details(name,age=0):
    return f"Student name is: {name}\nAge is: {age}" # function closed
    print("Hello world")

obj1 = student_details("Rahul")  # right position Note: Always follows the position of parameter
print(obj1)
````

````python
def student_details(name,age=18):
    return f"Student name is: {name}\nAge is: {age}" # function closed
    print("Hello world")

obj1 = student_details("Rahul",)  # right position Note: Always follows the position of parameter
print(obj1)
````

### 7. Scope of Varaibles

* 1. Global Varaible
* 2. Local Varaible

````python
x = 10  # Global Varaible

def testVariable():
    return x  # varaible calling inside function
testVariable()
````

````python
print(x)  # x is global varaible
````

````python
def testVariable():
    x1 = 10  # local Varaible
    return x1  # varaible calling inside function
testVariable()
````

````python
print(x1)
````

````python
def testVariable():
    global x2  # you are declare x2 as global
    x2 = 10  # global Varaible
    return x2  # varaible calling inside function
testVariable()
````

````python
print(x2)
````

````python
print(x2)
````

### 9. Nested Functions

function inside another function


````python
def outerFunc():
    def innerFunc():
        print("Inner Function is calling")
        def inner2():
            print("Hello world")
        inner2()
    innerFunc()  # calling of inner function

outerFunc()  #
````

## 11. Docstrings



````python
def add1(a,b):
    """
    This function returns the sum of two numbers.This is called multi-line comment
    """
    return a+b

add1(10,12)
````

````python
def add1(a,b):
    """
    This function returns the sum of two numbers.This is called multi-line comment
    """
    return a+b


s1 = add1(10,12)
print(add1.__doc__)  # docstring
print(s1)
````

````python
def add1(a,b):
    """
    This function returns the sum of two numbers.This is called multi-line comment
    """
    return a+b


s1 = add1(10,12)
print(s1.__doc__)  # docstring
print(s1)
````
