```python
# 18. Read a file and count the number of vowels.

filename = r"C:\Users\Admin\Downloads\nursery_rhymes.txt"

count = 0

with open(filename, "r") as file:
    for line in file:
        for ch in line:
            if ch.lower() in "aeiou":
                count += 1

print("Number of vowels:", count)
```

```python
# 19. Read a file and count the number of digits.

filename = r"C:\Users\Admin\OneDrive\Desktop\The sun.txt"

count = 0

with open(filename, "r") as file:
    for line in file:
        for ch in line:
            if ch.isdigit():
                count += 1

print("Number of digits:", count)
```

```python
# 23. Read a file and reverse the order of all lines.

filename = r"C:\Users\Admin\OneDrive\Desktop\The sun.txt"

try:
    with open(filename, "r") as file:
        lines = file.readlines()

    print("Lines in Reverse Order:")
    for line in reversed(lines):
        print(line, end="")

except FileNotFoundError:
    print("File not found!")
```

```python
# 24. Read a file and print its contents in reverse character order.

filename = r"C:\Users\Admin\OneDrive\Desktop\The sun.txt"

try:
    with open(filename, "r") as file:
        text = file.read()

    print("File contents in reverse character order:")
    print(text[::-1])

except FileNotFoundError:
    print("File not found!")
```

```python
# 28. Read a JSON file and display all key-value pairs.

import json

try:
    with open(r"C:\Users\Admin\Downloads\sample1 (1).json") as file:
        data = json.load(file)

    print("Key-Value Pairs:")
    for key, value in data.items():
        print(key, ":", value)

except FileNotFoundError:
    print("File not found!")
except json.JSONDecodeError:
    print("Invalid JSON file!")
```

```python
# 31. Read a log file and count the number of ERROR, WARNING, and INFO messages.

file = open(r"C:\Users\Admin\OneDrive\Desktop\The sun.txt", "r")

error_count = 0
warning_count = 0
info_count = 0

for line in file:
    if "ERROR" in line:
        error_count += 1
    elif "WARNING" in line:
        warning_count += 1
    elif "INFO" in line:
        info_count += 1

file.close()

print("Number of ERROR messages:", error_count)
print("Number of WARNING messages:", warning_count)
print("Number of INFO messages:", info_count)
```
