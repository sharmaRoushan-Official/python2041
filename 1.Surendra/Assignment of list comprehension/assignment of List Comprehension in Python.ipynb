# Assignment of List Comprehension in Python

```python
# 1. Generate a list of squares from 1 to 10 using list comprehension.

[i**2 for i in range(1,10)]
```

```python
# 2. Create a list of even numbers from 1 to 50 using list comprehension.

[num for num in range(1, 51) if num % 2 == 0]
```

```python
# 3. Convert the list of strings to uppercase using list comprehension: words = ['cat', 'dog', 'lion']

words = ['cat', 'dog', 'lion']
[word.upper() for word in words]
```

```python
# 4. Get a list of numbers divisible by both 3 and 5 from 1 to 100.

[num for num in range(1, 101) if num % 3 == 0 and num % 5 == 0]
```

```python
# 5. Extract all vowels from the string using list comprehension: text = "Comprehension"

text = "Comprehension"
[char for char in text if char.lower() in 'aeiou']
```

```python
# 6. Given a matrix: matrix = [[1,2,3],[4,5,6],[7,8,9]] Use list comprehension to flatten it into a single list.

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
[num for row in matrix for num in row]
```

```python
# 7. Given a list of words, extract words that contain the letter ‘e’: words = ['tree', 'book', 'python', 'sky']

words = ['tree', 'book', 'python', 'sky']
[word for word in words if 'e' in word]
```

```python
# 8. Create a list of all 3-digit numbers that are palindromes using list comprehension.

[num for num in range(100, 1000) if str(num) == str(num)[::-1]]
```

```python
# 9. From a list of numbers, replace odd numbers with 'ODD' and keep even numbers as is: nums = [1, 2, 3, 4, 5, 6]

nums = [1, 2, 3, 4, 5, 6]
['ODD' if num % 2 != 0 else num for num in nums]
```

```python
# 10. Using nested list comprehension, generate a 3x3 identity matrix.

[[1 if i == j else 0 for j in range(3)] for i in range(3)]
```
