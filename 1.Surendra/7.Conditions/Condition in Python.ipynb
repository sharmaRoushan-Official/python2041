### 13. Advance (Case)

### match and case

```python
day = int(input("Enter any value from 1-7:"))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid")
```

```python
### 15. Create a simple calculater that performs basic arthmetic operations based on two user input
#(^,*,/,-,+)

num1 = float(input("Enter the first number:"))
num2 = float(input("Enter the second number"))
operator = input("Enter the arthmetic operator:")

if operator == "^":
    print("Result", num1**num2)
elif operator == "*":
    print("Result", num1*num2)
elif operator == "/":
    print("Result", num1/num2)
elif operator == "-":
    print("Result", num1-num2)
elif operator == "+":
    print("Result", num1+num2)
else:
    print("Invaild Operator")
```
