```python
# 5. Create a parent class Shape with a method area(). Derive classes Circle and Rectangle. Implement the area method in each subclass.

import math

class Shape:
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


c = Circle(5)
r = Rectangle(10, 4)

print("Area of Circle:", round(c.area(), 2))
print("Area of Rectangle:", r.area())
```

```python
# 8. Demonstrate *operator overloading* by overloading the + operator for a class Point(x, y) to add coordinates.

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print("(", self.x, ",", self.y, ")")

p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

print("First Point:")
p1.display()

print("Second Point:")
p2.display()

print("Sum of Points:")
p3.display()
```

### *Encapsulation (11–15)*


```python
# 11. Create a class Account with private attributes __balance. Provide methods deposit() and withdraw() to modify it safely.

class Account:
    def __init__(self, balance):
        self.__balance = balance  

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited:", amount)
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount!")
        elif amount <= self.__balance:
            self.__balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance!")

    def display_balance(self):
        print("Current Balance:", self.__balance)

acc = Account(10000)

acc.display_balance()
acc.deposit(3000)
acc.withdraw(1500)
acc.display_balance()
```

```python
# 12. Write a class with a private method __calculate_salary() and a public method get_salary() that internally calls the private method.

class Employee:
    def __init__(self, basic_salary, bonus):
        self.basic_salary = basic_salary
        self.bonus = bonus

    def __calculate_salary(self): 
        return self.basic_salary + self.bonus

    def get_salary(self): 
        return self.__calculate_salary()

emp = Employee(20000, 4000)

print("Total Salary:", emp.get_salary())
```

```python
# 16. Create an abstract class Vehicle with an abstract method start(). Implement it in subclasses Car and Bike.

from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):
    def start(self):
        print("Car starts with a key.")


class Bike(Vehicle):
    def start(self):
        print("Bike starts with a self-start button.")


car = Car()
bike = Bike()

car.start()
bike.start()
```

```python
# 17. Create an abstract class Shape with abstract methods area() and perimeter(). Implement these in Circle and Square.

from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

    def perimeter(self):
        return 4 * self.side

c = Circle(12)
s = Square(14)

print("Circle Area:", round(c.area(), 2))
print("Circle Perimeter:", round(c.perimeter(), 2))

print("Square Area:", s.area())
print("Square Perimeter:", s.perimeter())
```

```python
# 18. Define an abstract class FileHandler with methods read() and write(). Create child classes TextFileHandler and BinaryFileHandler.

from abc import ABC, abstractmethod

class FileHandler(ABC):

    @abstractmethod
    def read(self, filename):
        pass

    @abstractmethod
    def write(self, filename, data):
        pass

class TextFileHandler(FileHandler):

    def read(self, filename):
        with open(filename, "r") as file:
            return file.read()

    def write(self, filename, data):
        with open(filename, "w") as file:
            file.write(data)

class BinaryFileHandler(FileHandler):

    def read(self, filename):
        with open(filename, "rb") as file:
            return file.read()

    def write(self, filename, data):
        with open(filename, "wb") as file:
            file.write(data)

text_handler = TextFileHandler()
text_handler.write("sample.txt", "Hello, Python!")
print("Text File Content:", text_handler.read("sample.txt"))

binary_handler = BinaryFileHandler()
binary_handler.write("sample.bin", b"Hello, Binary!")
print("Binary File Content:", binary_handler.read("sample.bin"))
```

```python
# 19. Create an interface-like abstract class Remote with methods power_on() and power_off(). Implement it for TV and AC classes.

from abc import ABC, abstractmethod

class Remote(ABC):

    @abstractmethod
    def power_on(self):
        pass

    @abstractmethod
    def power_off(self):
        pass

class TV(Remote):

    def power_on(self):
        print("TV is turned ON")

    def power_off(self):
        print("TV is turned OFF")

class AC(Remote):

    def power_on(self):
        print("AC is turned ON")

    def power_off(self):
        print("AC is turned OFF")

tv = TV()
ac = AC()

tv.power_on()
tv.power_off()

ac.power_on()
ac.power_off()
```

```python
# 20. Implement an abstract class Payment with method make_payment(). Create subclasses CardPayment, UPIPayment, and WalletPayment with different implementations.

from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def make_payment(self, amount):
        pass

class CardPayment(Payment):

    def make_payment(self, amount):
        print(f"Payment of ₹{amount} made using Card.")

class UPIPayment(Payment):

    def make_payment(self, amount):
        print(f"Payment of ₹{amount} made using UPI.")

class WalletPayment(Payment):

    def make_payment(self, amount):
        print(f"Payment of ₹{amount} made using Wallet.")
        
card = CardPayment()
upi = UPIPayment()
wallet = WalletPayment()

card.make_payment(1000)
upi.make_payment(500)
wallet.make_payment(200)
```
