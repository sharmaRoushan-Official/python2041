# Assignment of Lambda

```python
# 1. Write a lambda function to calculate the square of a number.

(lambda a:a*a)(323)
```

```python
# 2. Use a lambda function with map() to add 5 to each number in the list nums = [1, 2, 3, 4]

numbers = [1,2,3,4]
list(map(lambda a:a+5,numbers))
```

```python
# 3. Use filter() with a lambda to find even numbers from a list nums = [5, 8, 11, 14, 17, 20]

nums = [5, 8, 11, 14, 17, 20]
list(filter(lambda a:a%2==0,nums))
```

```python
# 4. Use reduce() with a lambda to compute the product of all elements in a list nums = [2, 3, 4, 5]

nums = [2, 3, 4, 5]

from functools import reduce

reduce(lambda a,b:a*b,nums)
```

```python
# 5. Sort the following list of tuples by the second value using a lambda function: data = [(1, 3), (2, 1), (4, 2)]

data = [(1, 3), (2, 1), (4, 2)]
sorted(data, key= lambda x:x[1])
```

```python
# Write a lambda that checks if a string starts with 'A'.

starts_with_a = lambda x: x.startswith('A')
```

```python
print(starts_with_a("Apple"))  
print(starts_with_a("Banana")) 
```

```python
# 7. Use a lambda function with map() to convert a list of Celsius temperatures to Fahrenheit:temps = [0, 20, 37, 100]

temps = [0, 20, 37, 100]
list(map(lambda a:(a*9/5)+32,temps))
```

```python
# 8. Given a list of strings, filter all strings with length > 3 using a lambda:words = ['pen', 'notebook', 'apple', 'go']

words = ['pen', 'notebook', 'apple', 'go']

result = list(filter(lambda word: len(word) > 3, words))

print(result)
```

```python
# 9. Create a lambda that takes two arguments and returns their average.

(lambda a,b:(a + b) / 2)(644,86688)
```

```python
# 10. Use a lambda and map() to capitalize the first letter of each word in a list:words = ['python', 'lambda', 'function']

words = ['python', 'lambda', 'function']
list(map(lambda word: word.capitalize(), words))
```
