### what is for loop?

A for loop is used to iterate over a sequence(such as list, tuple, string, dictionary, set, range) and execute a block of repeatedly.

Flow:

1. Python takes the first item first item from the sequence.
2. Assigns it to the loop variable.
3. Executes the block of code.
4. Move the next item.
5. Repeats until all itmes are processed.

syntax:

for variable in  iterable:
    
    print(variable)`````

``````python
fruits = ["apple","banana","mango"]

for fruit in fruits:
    print(fruit) # apple banana mango
``````

### Explanation

|Iteraation|Fruit value|
|----|----|
|1|apple|
|2|banana|
|3|mango|

used for matrices, pattern tables.

``````python
for i in range(3): # outer loop : 0,1,2
    for j in range(2): # inner loop : 0,1
        print(i,j) # statement
``````

``````python
for i in range(1,4): # outer loop : 0,1,2,3
    for j in range(3): # inner loop : 0,1,2
        print(i,j) # statement
``````

### Different between for loop and while loop
-------

|for loop|while loop|
|----|-----|
|Iterates over sequence|Runs based on conditions|
|Known iteration|Unknown iterations|
|Simpler|More fiexble|

### While Loop:-

A while loop repeatedly executes a block of code as long as condition is true.

* There is not range keyword in while loop.
* You need to start and stop their values usning condtion.
* You need to increment their value also.
* for loop = known number of iteration.
* while loop = Unknown number of iterations, depends on condition.


### What is a while loop?

A while loop checks a condition before each iteration

syntax:

while condition:

code block

``````python
# 1-11 using while loop

i = 1 # intial value

j = 11 # final value

while i<j: # 1<11 2<11 3<11 4<11 5<11 6<11 7<11 8<11 9<11 10<11 (11<11--->False)
    print(i) # 1 2 3 4 5 6 7  9 10 
    i+=1
``````

``````python
# tablenum of 5 using while loop
num = 5

i = 1

j = 11

while i<j:
    print(f"{i}x{num}={i*num}")
    i+=1
``````

### Nested While Loop

``````python
i = 1
j = 3

while i<=j: # 1 2 3
    a = 1
    b = 5
    while a<=b: # 1 2 3 4 5
        print(i,a)
        a+=1
    i+=1
``````

``````python
i = 1

j = 11 

while i<j:
    print(f"=====Table of {i}=====")
    a = 1 
    b = 11
    while a<b:
        print(i,'x',a,'=',i*a)
        a+=1
    print()
    i+=1
``````

### Break and continue in while loop 

``````python
# break
# 1-11

i = 1
j = 11

while i<j: 
    if i == 5:
        break # stop value 4
    print(i)
    i+=1
``````

``````python
# 1-11

i = 1
j = 11

while i<j:
    print(i)
    if i == 5: # stop value 5
        break
    i+=1
``````

``````python
# continue

# 1-11

i = 1
j = 11

while i<j: # 1<11 2<11 3<11 4<11 5<11
    i+=1 # 2
    print(i)
    if i == 5: # continue line upto 11 the  stop value
        continue
``````

# pattern using while loop 

``````python
# 1. Increasing Star Pattern
# *
# **
# ***
# ****
# *****

i = 1

while i <= 5:
    j = 1
    while j <= i:
        print("*", end="")
        j += 1
    print()
    i += 1
``````

``````python
i = 5

while i >= 1:
    j = 1
    while j <= i:
        print("*", end="")
        j += 1
    print()
    i -= 1
``````

``````python
# 3. Square Star Pattern
# *****
# *****
# *****
# *****
# *****

i = 1

while i <= 5:
    j = 1
    while j <= 5:
        print("*", end="")
        j += 1
    print()
    i += 1
``````

``````python
# 4. Increasing Number Pattern
# 1
# 12
# 123
# 1234
# 12345

i = 1

while i <= 5:
    j = 1
    while j <= i:
        print(j, end="")
        j += 1
    print()
    i += 1
``````

``````python
# 5. Continuous Number Pattern
# 1
# 23
# 456
# 78910

i = 1
num = 1

while i <= 4:
    j = 1
    while j <= i:
        print(num, end="")
        num += 1
        j += 1
    print()
    i += 1
``````

``````python
# 6. Repeated Number Pattern
# 1
# 22
# 333
# 4444
# 55555

i = 1

while i <= 5:
    j = 1
    while j <= i:
        print(i, end="")
        j += 1
    print()
    i += 1
``````
