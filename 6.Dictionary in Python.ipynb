## 7. Dictionary in Python

----
#### Q 1. What is Dictionary in Python?

A `dictionary` is a collectio of key:values pairs

* Each key is unique
* Each key maps to a value.
* Dicationary are mutable(Unhashable type/Changeable)
* Dictionary bracket is Curly bracket {}

```python
d1 = {}
print(d1)
print(type(d1))
```

```python
x1 = set()
print(x1)
print(type(x1))
```

```python
student = {
    "name":"Arnav",
    "age":17,
    "course":"Python"
}
print(student)
print(type(student))
```

## 2. Why We Use Dictionaries

Use Dictionary when data has a relationship between key and values.

Example Uses:

-----

|Key|Value|
|----|---|
|name|"Arnav"|
|age|17|
|course|Python|


```python
employee = {"id":101,'name':"Amit","salary":50000}
print(employee)
print(type(employee))
```

## 2. Key must be Immutable
----

Allowed:

* string
* number
* tuple

---
Not Allowed:

* List
* Dictionary
* set

```python
data = {
    1:"Python",
    "name":"Arnav",
    (10,20):"Tuple Key",
    True: "Boolean",
    False:"Hello"
}
data
```

```python
data = {
    1:"Python",
    "name":"Arnav",
    (10,20):"Tuple Key",
    True: "Boolean",
    False:"Hello",
    [10,20]:"List Data"
}
data
```
