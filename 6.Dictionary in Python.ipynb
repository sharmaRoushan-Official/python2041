## 7. Dictionary in Python

----
#### Q 1. What is Dictionary in Python?

A `dictionary` is a collectio of key:values pairs

* Each key is unique
* Each key maps to a value.
* Dicationary are mutable(Unhashable type/Changeable)
* Dictionary bracket is Curly bracket {}

```python
d1 = {}
print(d1)
print(type(d1))
```

```python
x1 = set()
print(x1)
print(type(x1))
```

```python
student = {
    "name":"Arnav",
    "age":17,
    "course":"Python"
}
print(student)
print(type(student))
```

### 11. Looping Through Dictionary

```python
student
```

```python
for key in student:
    print(key)
```

### 12. Nested Dictionary

----

Dictionary inside another dictionary


```python
student = {
    "student1":{
        "name":"Arnav",
        "age":17
    },
    "student2":{
        "name":"Rahul",
        'age':26
    },
    "student3":{
        "name":"Ankita",
        "age":45
    }
}
print(student)
```
