### 2. Class

A `class` is a blueprint or template used to create any object.

syntax

```python
class student:
    pass
```

#### Key-Points

* Defined using class keyword
* Can contains variables and method
* Naming Convention: PascalCase

```python
class Student:  # Student is the name of template
    def display(self):  # display is the method
        return "Welcome to Python OOPs"  # method statement

obj1 = Student()
print(obj1.display())
```

### 3. Object

An Object is an instance of a class

syntax:

```python
s1 = Student()
print(s1.display())
```

```python
print(obj1.display())
print(s1.display())
```

Keypoint

* memory allocated when object is created.
* can access class method and varibles.

### 4. Self Keyword

Define

self refers to the current object of the class.

### Why Required?
* To access instance varaibes and methods
* Differetiate instance variables from local varaibes.

```python
class Demo:
    def show(self):
        print("Self representes current object")
```

```python
x1 = Demo()
x1.show()
```

#### 5. Constructor `(__init__)`

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

x1 = Student("Surendra",35)  # 
print(x1.name)
print(x1.age)
```

```python
x2 = Student("Tanaz",45)
print(x2.name)
print(x2.age)
```
