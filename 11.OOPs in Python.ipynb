### a). Single/Simple Inheritance

----
Meaning:
* In `Single Inheritance`, a child inherits from only one parent class.

📌Structure:

`Parent ➺ Child`

#### Why It is used:
* Code reusability
* Child can use parent class property and method.
* Easy to understand and maintain


### Real-Life Example

* Parent - Vehicle
* Child- Car

A Car is a Vehicle


Python Example


````python
class Vehicle: # parent
    def start(self):
        print("Vehicle Started.")

class Car(Vehicle): # child / derived class
    def drive(self):
        print("Car is driving")
        
# obj1 = Vehicle()
# obj1.start()

# obj = Car()
# obj.drive()

obj = Car()
obj.drive()
obj.start()
````

Note: 

* Single Inheritance is most basic and safest inheritance. It avoids complexity and ambiguity.


_Ambiguity in Object-Oriented Programming (OOP) occurs when the compiler or interpreter cannot determine which member function, variable, or path to pick because two or more options match the same call._

### b). Multiple Inheritance

Meaning:

In `Multiple Inheritance`, a child class inherits from more than one parent class.

📌 Structore

```
         Parent1           Parent2
                \         /
                 \       /
                  \     /
                   child
```
### Why it is used?
* When one class needs feature from multiple class.
* Helps to combine functionalities.


#### Real-Life Example

* Parent1: Camera
* Parent2: Phone
* Child: Smart Phone

SmartPhine is both a Phone and a Camera

Python Example

````python
class Phone:  # parent1
    def call(self):
        return "Calling......"

class Camera:  # parent2
    def click_photo(self):
        return "Photo Clicked"


class SmartPhone(Camera,Phone):  # Child/Derived class
    pass


obj = SmartPhone()
print(obj.call())
print(obj.click_photo())
````

Note:

Python Support Multiple Inheritance, You can check order of method called Method Resolution Order(mro) to avoid confusion.

````python
print(SmartPhone.mro())
````

### 3. Multilevel Inheritance

Meaning

In `Multilevel Inheritance`, inheritance happens in levels.

📌 Structure

`Grandparent➺Parent➺Child`

### Why it is used?

* To represent hierarchy in real world.
* Step-by-step inheritance of feature.


### Real-Life Example
* Grandparent: Animal
* Parent: Mammal
* Child: Dog

##### Dog is a Mammal, Mammal is an Animal

````python
class Animal: # grandparent
    def eat(self):
        return "Eating...."


class Mammal(Animal): # parent
    def walk(self):
        return "Walking...."

class Dog(Mammal): # child
    def bark(self):
        return "Barking...."


d = Dog()
print(d.bark()) # bark is the property of child
print(d.eat()) # eat is the property of Animal
print(d.walk()) # walk is the property of mammal
````

Note:

Child class gets access to all previous levels method.

    

````python
print(Dog.mro())
````

#### 4. Hierarchical Inheritance

Meaning:

In `Hierarchical Inheritance`,Multiple child classes inherits from a single parent class.


📌Structure

```
         Parent1
        /   |   \
       /    |    \
      /     |     \
  child1  child2  child3
```
#### Real-Life Example

* Parent: Employee
* Children:
  * Manager
  * Developer
  * Tester
* All are employees but have different roles.

##### Python Example


````python
class Employee:
    def work(self):
        return "Employee is working...."

class Manager(Employee):
    def manage(self):
        return "Manager is managing work...."

class Developer(Employee):
    def code(self):
        return "Developer is coding...."

class Tester(Employee):
    def test(self):
        return "Tester is Testing....."


m = Manager()
print(m.work()) # parent property
print(m.manage()) # child property

d = Developer() 
print(d.work()) # parent property
print(d.code()) # child property


t = Tester()
print(t.work()) # parent property
print(t.test()) # child property
````

````python
print(Tester.mro())
````

#### 5. Hybrid Inheritance:

Meaning:

Hybrid Inheritance is combination of tow or more inheritance types.

It may Include:

* Multiple + Multilevel
* Hierarchical+ Multiple

📌 Structure:

```
            A
           / \
          /   \
         B     C
          \   /
           \ /
            D
```
Here:

* B and C inherits from A(Hierarchial Inheritance)
* D inherits from B and C(Multiple Inheritance)

  `or`
  
* D inherits from B inherits from A (Multilevel Inheritance)
* C inherts from A and D (Multiple Inheritance)

### Why It is Used?

* When a system requires complex relationships.
* Common in large framework and entrerprises project.


#### Python Example

````python
class A:
    def showA(self):
        return "Class A"

class B(A):
    def showB(self):
        return "Class B"

class C(A):
    def showC(self):
        return "Class C"

class D(B,C):
    def showD(self):
        return "Class D"

obj = D()
print(obj.showA())
print(obj.showB())
print(obj.showC())
print(obj.showD())
````

````python
print(D.mro())
````

````python
class A:
    def showA(self):
        return "Class A"

class B(A):
    def showB(self):
        return "Class B"

class C(A,D):
    def showC(self):
        return "Class C"

class D(B):
    def showD(self):
        return "Class D"


obj = C()
print(obj.showA())
print(obj.showB())
print(obj.showC())
print(obj.showD())
````

Note:

Hybrid Inheritance is powerful but can cause confusion due to :

* Method Conflicts
* Ambioguity

That's by Python uses mro(Method Resoluction Order.)

````python
print(C.mro())
````

### Encapsulation using @property

````python
class Employee:
    def __init__(self):
        self.__salary = 0

    @property  # value return 
    def salary(self):
        return self.__salary  # display salary on the screen

    @salary.setter # value set
    def salary(self,amount):
        if amount>0:
            self.__salary = amount
            print(f"Salary Updates Successfully:{self.__salary}")
        else:
            print("Invalid Salary")


emp = Employee()
emp.salary = 4000 # setter
````

````python
emp.salary
````

````python
emp.salary
````

````python
emp.salary = 50000
````

### 4. Data Abstraction

Meaning of Data Abstraction

Data Abstraction means hiding internal implemenatation details and showing only essential feature to the user.

📌Simple Definition
> Whan an object does is shown, how it does it hidden.

### Why Data abstraction is needed

#### ❌ Without Abstraction
* User sees unnecessary details
* Code become complex
* Difficult to maintain
* Higher chances to misuse
#### ✅With Abstraction
* Only required feature exposed
* Internal logic hidden
* Easy to use and maintain
* Improve security

#### Real-Life Example

Driving a Car

-----
* You use:
  * Sterring
  * Accelerator
  * Break
* You don't know engine internal

This is Data Abstraction

------
##### Data Abstraction VS Encapsulation

----
|`Feature`|`Encapsulation`|`Data Abstraction`|
|-----|-----|-----|

|`Shows`|How data is protected|What functionality is provided|
|`Achieved by`|Access Modifiers|Abstract base classes/interface|
|`Logic`|Public,Protected,Private|Module:abc, abstractmethod|

Note:

* Encapsulation = How data is stored
* Abstraction = What user can do

-----
### How Abstraction is Achieved in Python

#### Python provides:
  1. Abstract Base Classes(ABC)
  2. Abstract Method [Built-in Decorator]


1. Abstract Classe(Core Concept)

   📌 An Abstract class:

   * Cannnot be instaniated
   * Contains abstract method
   * Acts as a blueprint
------
2. Abstract Method

   📌 Abstract Method

   * Has declaration only
   * No implementation
   * Must be implementaed by child class

````python
# Data Abstraction 
from abc import ABC,abstractmethod
class Vehicle(ABC):  # base class
    @abstractmethod
    def start(self):
        print("Vehicle Started")
    @abstractmethod
    def abc(self):
        # print(asljfaflijawslkjfsqa)
        pass
````

````python
# v = Vehicle()
# v.start()
````

````python
class Car(Vehicle):  # child class or derived class
    def carEngine(self):
        print("Car has BS-IV engine-2019 Model")

    def start(self):
        print("Car is started.")

    def abc(self):
        pass

    def sterring(self):
        print("Car has sterring")

c = Car()
c.carEngine()
c.sterring()
c.start()
````

#### Rules of Data Abstraction
* Abstract class must inherits ABC(Base Class)
* Abstract method must use @abstractmethod
* Child class must implement abstract method
* Abstract class object cann't be creatd.


### When to use Abstraction:
Use Abstraction when:
* Common functionaliryt exists
* Multiple Implementation needed
* System design is complex
* Freamework or large application

Notes:

* Abstraction hides implemenation details
* Achieved using abstract class
* Show essential features only.
* Improve security ans scalability
