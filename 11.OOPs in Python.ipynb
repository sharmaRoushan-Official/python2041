### a). Single/Simple Inheritance

----
Meaning:
* In `Single Inheritance`, a child inherits from only one parent class.

📌Structure:

`Parent ➺ Child`

#### Why It is used:
* Code reusability
* Child can use parent class property and method.
* Easy to understand and maintain


### Real-Life Example

* Parent - Vehicle
* Child- Car

A Car is a Vehicle


Python Example


````python
class Vehicle: # parent
    def start(self):
        print("Vehicle Started.")

class Car(Vehicle): # child / derived class
    def drive(self):
        print("Car is driving")
        
# obj1 = Vehicle()
# obj1.start()

# obj = Car()
# obj.drive()

obj = Car()
obj.drive()
obj.start()
````

Note: 

* Single Inheritance is most basic and safest inheritance. It avoids complexity and ambiguity.


_Ambiguity in Object-Oriented Programming (OOP) occurs when the compiler or interpreter cannot determine which member function, variable, or path to pick because two or more options match the same call._
