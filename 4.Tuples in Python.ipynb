## 4. Tuples in Python


#### Q. What is a Tuples?

A `tuple` is a collection data type in Python used to store multiple values in single variable.

or 

A `tuple` is a immutable type of list.



#### Key Characteristics:

* Ordered collection
* Immutable(cannot be changed after creation)
* Allows duplication values
* Can store different data types
* Faster than lists
* Written using round brackets ()


```python
numbers = (10,20,30,40)
print(numbers)
```

```python
type(numbers)
```

### 5. Accessing Tuple Elements

Tuple elements are accessed using indexing

1. Positive Indexing
2. Negative Indexing

```python
fruits = ("apple","banana","mango")
print(fruits)
```

```python
# 1.Positive Indexing

print(fruits[0])
print(type(fruits[0]))

print(fruits[1])
print(type(fruits[1]))

print(fruits[2])
print(type(fruits[2]))
```

```python
# 2. Negative Indexing 

data = ("John",25,75.5,True,False)
print(data)

print(data[-1])
print(type(data[-1]))

print(data[-2])
print(type(data[-2]))

print(data[-3])
print(type(data[-3]))

print(data[-4])
print(type(data[-4]))

print(data[-5])
print(type(data[-5]))
```

### 6. Slicing of Elements:

### 7. Tuple Immutability

Tuple cannot be modified after creation



```python
numbers = (10,20,30)
print(numbers)
```

```python
numbers[0] = "hello world"
print(numbers)
```

### 8. Changing Tuple Elements:

Since tuples are immutable, convert into list.

```python
numbers = (10,20,30)
print(numbers)
print(type(numbers))

temp_list = list(numbers)  # tuple converted into list.
print(temp_list)
print(type(temp_list))

temp_list[0] = "Hello world"
temp_list


tup = tuple(temp_list)
print(tup)
print(type(tup))
```

### 20. Advantages of Tuples

21. Tuple vs List

|`Feature`|Tuple|List|
|----|----|----|
|`Brakets`|()|[]|
|`Multable`|No|Yes|
|`Immutable`|Yes|No|
|`Speed`|Faster|Slower|
|`Memory`|Less|More|

Example:

```python
student = ("Ram",21,"Computer Science",85)
print(student)
```

```python
name,age,course,marks = student
print("Name:",name)
print("Age:",age)
print("Course:",course)
print("Marks:",marks)
```

## 23. When to Use Tuple:

---
Use tuple when:

* Data should not change
* Fixed records(Coordinate,database-row,column)
* Returning multiple values from function 


## 24. Tuple Performance Advantage

Tuple are faster than lists becuse they are immutable and consumes less primary memeory.
