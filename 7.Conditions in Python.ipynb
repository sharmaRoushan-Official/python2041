## 2. Base if Statment

Syntax:

if Condition:

    # Code execute if condition is True
    # statement

```python
age = 16

if age>=18:  # age>18: True/False
    print("You Can Vote")  # True Block
```

```python
age = 19

if age>=18:  # age>18: True/False
    print("You Can Vote")  # True Block

else:
    print("You cann't vote")
```

# Key Points:

* Condition must return True/False
* Indentation is mandatory
* No paretheses is required

```python
age = 19

if (age>=18):  # age>18: True/False
    print("You Can Vote")  # True Block

else:
    print("You cann't vote")
```

## 4. if-elif-else

1- Monday
2 - Tuesday
3 - Wednesday
4 - Thursday
5- Friday
6- Saturday
7- Sunday

```python
week = 10

if week == 1:
    print("Monday")
elif week == 2:  # 2==2
    print("Tuesday")

elif week == 3:
    print("Wednesday")
elif week == 4:
    print("Thursday")
elif week == 5:
    print("Friday")
elif week == 6:
    print("Saturday")

else:
    print("Sunday")
```

```python
week = 8

if week == 1:
    print("Monday")
elif week == 2:  # 2==2
    print("Tuesday")

elif week == 3:
    print("Wednesday")
elif week == 4:
    print("Thursday")
elif week == 5:
    print("Friday")
elif week == 6:
    print("Saturday")
elif week == 7:
    print("Sunday")

else:
    print("No Week Found")
```
