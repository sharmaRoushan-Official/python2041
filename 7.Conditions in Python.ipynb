## 2. Base if Statment

Syntax:

if Condition:

    # Code execute if condition is True
    # statement

```python
age = 16

if age>=18:  # age>18: True/False
    print("You Can Vote")  # True Block
```

```python
age = 19

if age>=18:  # age>18: True/False
    print("You Can Vote")  # True Block

else:
    print("You cann't vote")
```

* Condition must return True/False

### 5. Nested if Statements

```python
age = int(input("Enter your age: "))

citizen = True
if age>0 and age<=100: # 99>0  and 99<=100
    if age>=18:  # 19>=18
        if citizen:
            print("Elegible to vote")
        else:
            print("Not Elegible")
    else:
        print("Not Elegible to vote")
else:
    print("Invalid Age")
```

### 13. Advanced (Case)

### match and case

```python
day = int(input("Enter any value from 1-7: "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid")
    
```
