### 13. Writing Files in Python

Q. What is Writing Files?

Writing Fliles means storing data permanently in file from python program.


📌 Used for:

* Saving User data
* Logs
* Reports
* Exporting result
* Backup files

### File Writing Modes

|Mode|Meaning|
|----|----|
|`w`|Write(Created new/Overwrite)|
|`a`|Append(data at end)|
|`w+`|Write+Read|
|`a+`|Append and Read|
|`x`|Create if already exists it show error|

#### There are only three steps to creating files
##### Step-by-step process
1. Open the file
2. Write data in the file
3. Close the file

```python
fileName = r"C:\Users\User\Desktop\surendra.txt"

# step-1
file = open(fileName,'w')

# step-2

file.write("Myself Surendra\n")
file.write("I Live on the earth.")

# step-3
file.close()
```

```python
#### append() 

fileName = r"C:\Users\User\Desktop\surendra.txt"

# step-1
file = open(fileName,'a') # append

# step-2

file.write("\nMyself Surendra\n")
file.write("I Live on the earth.")

# step-3
file.close()
```
