### File Writing Modes

|Mode|Meaning|
|----|----|
|`w`|Write(Created new/Overwrite)|
|`a`|Append(data at end)|
|`w+`|Write+Read|
|`a+`|Append and Read|
|`x`|Create if already exists it show error|

#### There are only three steps to creating files
##### Step-by-step process
1. Open the file
2. Write data in the file
3. Close the file

```python
fileName = r"C:\Users\User\Desktop\surendra.txt"

# step-1
file = open(fileName,'w')

# step-2

file.write("Myself Surendra\n")
file.write("I Live on the earth.")

# step-3
file.close()
```

### 3. Best Practive: Writing Using With


```python
with open(r"D:\python2041\Surendra.txt","w") as file:
    file.write("Using With Statement\n")
    file.write("File auto closes.")
```

### 4. Writig User Input data to file 



```python
name = input("Enter your name: ")
age = input("Enter your age: ")

with open(r"D:\python2041\Surendra.txt",'w') as file:
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
```

```python
with open(r"D:\python2041\Surendra.txt",'a+') as file: # append data 
    name = input("Enter your name: ")
    age = input("Enter your age: ")
    file.write("-----------------------------\n")
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
    
```

### 5. Write + Read (w+ Mode)



```python
with open("Sample.txt","w+") as file:
    file.write("Hello world\n")
    file.seek(0)  # set cursor to 0
    print(file.read())
    file.seek(11)
    file.write("\nHello Python")
    file.seek(0)
    print(file.read())
```

### 6. Writing Using Loop(Log File)



```python
with open("log.txt",'a') as file:
    for i in range(3):
        file.write(f"Log entry:{i}\n")
```

### 7. Exceptional Hanlding While Writing File

```python
try:
    with open("data.txt",'x') as file:
        file.write("Hello world")
except Exception as e:
    print(e)
```

### PART-1: Reading + Writing Files(Combined)

Concept:

Most real application read data, process it, and write output.

Example Use Cases:

* Read Student marks -> Write result
* Read logs -> Write Reports
* Read input files -> Generate output file

Example: Read from one file and write another input file.txt


```python
with open("input.txt",'r') as infile:   # File Reading
    content = infile.read()
    # print(content)
with open("output.txt",'w') as outfile:
    outfile.write("Fruits Name:\n")
    outfile.write(content)
    print("Data Writes Successfully!!")
```

#### PART-2: CSV File Writing in Python

What is CSV?

* Comma Seperated Values
* Used in Excel, database, reports.

Required Library
`import csv`

```python
import csv
```

```python
with open("student.csv","w",newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["ID","Name","Marks"])
    writer.writerow([1,"Ram",99])
    writer.writerow([2,"Ramesh",85])
    writer.writerow([3,"Geeta",78])
```

Notes:

* newline= "" avoids blank lines
* CSV used in data science and backend
* Very Important for real projects.

```python
# import this library 

from openpyxl import Workbook
```

```python
wb = Workbook()
print(wb)  # check this workbook

sheet = wb.active
print(sheet)  # check this sheet

### Create column in excel file
sheet["A1"] = "ID"
sheet["B1"] = "Name"
sheet["C1"] = "Salary"

### Create Row in excel file [ Add data in excel file]
sheet.append([1,"Ram",90000])
sheet.append([2,"Ramesh",45000])
sheet.append([3,"Shyam",38000])

# give file name and save this file 
wb.save("employee.xlsx")
```
